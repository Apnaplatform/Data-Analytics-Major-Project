# Visualization

```python
sns.set_style('darkgrid')
plt.figure(figsize=(6,4))
plt.pie(df['Gender'].value_counts(), labels = df['Gender'].value_counts().index, autopct='%1.1f%%')
plt.title('Total Employee')
```

```python
colors = ['#ff9999', '#66b3ff', '#ffcc99', 'lightcoral', 'lightskyblue', 'gold']

counts = df["Gender"].value_counts()
explode= (0, 0.1)

counts.plot(kind='pie', fontsize=12, explode=explode, autopct='%1.1f%%', colors=colors)
plt.xlabel('Gender', weight="bold", fontsize=14, labelpad=20)
plt.axis('equal')
plt.legend(labels=counts.index, loc="best")
plt.show()
```

```python
department_counts = df['Department'].value_counts()
plt.figure(figsize=(20,6))
department_counts.plot(kind='bar', color ='darkblue')
plt.title('Total Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
Field_counts = df['EducationField'].value_counts()
plt.figure(figsize=(10, 6))
bars = Field_counts.plot(kind='bar', color=colors)

for bar, count in zip(bars.patches, Field_counts):
    plt.text(bar.get_x() + bar.get_width() / 2 - 0.1, bar.get_height() + 0.05, str(count), fontsize=10)

plt.title('Education Feild Distribution')
plt.xlabel('EducationField')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
Daily_counts = df['DailyRate'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(Daily_counts, labels=Daily_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Daily Rate Distribution')
plt.axis('equal')
plt.show()
```

```python
Attrition_counts = df['Attrition'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(Attrition_counts, labels=Attrition_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Attrition Distribution')
plt.axis('equal')
plt.show()
```

```python
age_group_counts = df['Age'].value_counts()
plt.figure(figsize=(10, 6))
bars = age_group_counts.plot(kind='bar', color=colors)

for bar, count in zip(bars.patches, age_group_counts):
    plt.text(bar.get_x() + bar.get_width() / 2 - 0.1, bar.get_height() + 0.05, str(count), fontsize=10)

plt.title('Age Group Distribution')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
pivot_df = df.pivot_table(index='Age', columns='Gender', aggfunc='size', fill_value=0)
ax = pivot_df.plot(kind='bar', color=['skyblue', 'lightcoral'], figsize=(10, 6))
ax.set_title('Age Group Distribution Based on Gender')
ax.set_xlabel('Age Group')
ax.set_ylabel('Count')
ax.legend(title='Gender')

plt.show()
```
